# Feature Engineering

```python
temp_cols = ['Avg_Temperature', 'Avg_Feels_Like_Temperature']

for col in temp_cols:
    data[col] = data[col].apply(lambda x: x - 273.15 if (x > 100) else x)

data.head()
```

```python
# Create a new column with log-transformed rain amount
data['Rain_Amount_log'] = np.log1p(data['Rain_Amount'])
```

```python
# Step 1: Classification based on rain amount
def classify_rain(mm):
    if mm == 0:
        return 'No Rain'
    elif mm <= 2.5:
        return 'Light'
    elif mm <= 10:
        return 'Moderate'
    else:
        return 'Heavy'

# Step 2: Apply the classifier
data['Rain_Class'] = data['Rain_Amount'].apply(classify_rain)

# Step 3: Encode as numbers (ordinal)
rain_mapping = {
    'No Rain': 0,
    'Light': 1,
    'Moderate': 2,
    'Heavy': 3
}

data['Rain_Class_Encoded'] = data['Rain_Class'].map(rain_mapping)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
data['Radiation_scaled'] = scaler.fit_transform(data[['Radiation']])
```

```python
# Convert degrees to radians
data['Wind_Direction_rad'] = np.radians(data['Wind_Direction'])

# Calculate u and v components
data['Wind_u'] = data['Wind_Speed'] * np.cos(data['Wind_Direction_rad'])
data['Wind_v'] = data['Wind_Speed'] * np.sin(data['Wind_Direction_rad'])

# Drop intermediate if you like
data.drop(columns=['Wind_Direction_rad'], inplace=True)
```

```python
to_scale = ['Temperature_Range', 'Evapotranspiration']
scaler = MinMaxScaler()
data[to_scale] = scaler.fit_transform(data[to_scale])
```

# Model training and testing

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from statsmodels.tsa.arima.model import ARIMA
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from sklearn.metrics import mean_squared_error

# Preprocessing: We assume 'Radiation_scaled' and 'Rain_Amount_log' are the processed versions
# Let's focus on the target columns
target_columns = ['Avg_Temperature', 'Radiation_scaled', 'Rain_Amount_log', 'Wind_Speed', 'Wind_Direction']

# Select relevant columns for modeling (drop unnecessary columns like 'ID', 'Year', etc.)
data_filtered = data[['Year', 'Month', 'Day', 'kingdom', 'latitude', 'longitude',
       'Avg_Temperature', 'Avg_Feels_Like_Temperature', 'Temperature_Range',
       'Feels_Like_Temperature_Range',
       'Rain_Duration', 'Wind_Speed', 'Wind_Direction', 'Evapotranspiration',
       'Rain_Amount_log', 'Rain_Class', 'Rain_Class_Encoded',
       'Radiation_scaled', 'Wind_u', 'Wind_v', 'Temp_Radiation',
       'Wind_Combined','kingdom_encoded'] + target_columns]

# Handle missing data or any other preprocessing (imputation, outlier handling, etc.)
# Ensure no NaN values in the columns to be used
data_filtered = data_filtered.dropna()

# Split the data into training and test sets (80% train, 20% test)
train_size = int(len(data_filtered) * 0.8)
train, test = data_filtered[:train_size], data_filtered[train_size:]

# Prepare the data for ARIMA and LSTM models
def create_dataset(data, time_step=1):
    X, y = [], []
    for i in range(len(data) - time_step):
        X.append(data[i:(i + time_step), :])  # Include all features as input
        y.append(data[i + time_step, :])  # The target variables (next step values)
    return np.array(X), np.array(y)

# Scaling data (MinMax scaling between 0 and 1 for LSTM)
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(data_filtered[target_columns].values)

# Set time_step for LSTM (the number of previous time steps to consider for prediction)
time_step = 30  # This can be adjusted based on your data and domain knowledge

# Prepare datasets for LSTM
X, y = create_dataset(scaled_data, time_step)
X = X.reshape(X.shape[0], X.shape[1], X.shape[2])  # Shape for LSTM [samples, time_step, features]

# Split into train and test for LSTM
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

# Initialize lists to store RMSE for each target column
hybrid_rmse = {}

for target_column in target_columns:
    print(f"Processing {target_column}...")

    # 1. ARIMA model - Forecast for each target column
    arima_model = ARIMA(train[target_column].values.flatten(), order=(5, 1, 0))  # Ensure we pass a univariate series
    arima_fit = arima_model.fit()
    arima_forecast = arima_fit.forecast(steps=len(test))
    
    # 2. LSTM Model - Prepare for LSTM on each target column
    target_index = target_columns.index(target_column)
    y_train_target = y_train[:, target_index]
    y_test_target = y_test[:, target_index]

    # Create the LSTM model
    lstm_model = Sequential()
    lstm_model.add(LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])))
    lstm_model.add(LSTM(units=50))
    lstm_model.add(Dense(units=1))  # Output layer for one target variable

    lstm_model.compile(optimizer='adam', loss='mean_squared_error')
    lstm_model.fit(X_train, y_train_target, epochs=20, batch_size=32, verbose=0)  # Training LSTM without verbose

    # LSTM Prediction on test data
    lstm_predictions = lstm_model.predict(X_test)

    # Hybrid Model (simple average of ARIMA and LSTM predictions)
    hybrid_predictions = (arima_forecast + lstm_predictions[:, 0]) / 2  # Average of ARIMA and LSTM predictions

    # Evaluate the performance of the hybrid model (RMSE for each target column)
    mse = mean_squared_error(y_test_target, hybrid_predictions)  # Evaluate RMSE for the current target column
    rmse = np.sqrt(mse)

    hybrid_rmse[target_column] = rmse
    print(f"Hybrid Model RMSE for {target_column}: {rmse}")

# Print RMSE for each target column
print("\nRMSE for each target column:")
for column, rmse_value in hybrid_rmse.items():
    print(f"{column}: {rmse_value}")
```
